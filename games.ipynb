Importamos los datasets

```python
import pandas as pd

ratings_df = pd.read_csv("datasets/game_ratings.csv")
metadata_df = pd.read_csv("datasets/games_metadata_5k.csv")

ratings_df.describe()
metadata_df.describe()

# promedio de ratings por juego
ratings_agg_df = ratings_df.groupby("game_id").agg(
    avg_user_rating=("rating", "mean"),
    num_ratings=("rating", "count")
).reset_index()

# Merge con metadata
df = metadata_df.merge(ratings_agg_df, on="game_id", how="left")

#Eliminamos columnas que no nos sirven
df = df.drop(columns = ['cover_image', 'game_link', 'metacritic_url', 'description'])

print(f"Dimensiones del dataset: {df.shape}")
print(f"Filas duplicadas: {df.duplicated().sum()}")
print(f"\nValores nulos por columna:")
print(df.isnull().sum())
print()
# Eliminar filas con nulos (son pocos, 36 máximo sobre 5000+)
df = df.dropna(subset=["genres", "platforms", "released"])

# Verificar que no quedaron nulos
print(df.isnull().sum())
print(df.shape)
df.info()
# Verificar


```
